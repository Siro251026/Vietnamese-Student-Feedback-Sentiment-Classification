```python
import pandas as pd

df = pd.read_csv("../dataset/UIT-VSFC_train.csv")
val =  pd.read_csv("../dataset/UIT-VSFC_valid.csv")
test = pd.read_csv("../dataset/UIT-VSFC_test.csv")

print(df.shape)
print(df.columns)
```

```python
label_counts = df["label"].value_counts()
label_ratio = df["label"].value_counts(normalize=True) * 100

print(label_counts)
print()
print(label_ratio.round(2))
```

```python
# Thông tin cấu trúc và chất lượng dữ liệu train
df.info()

print("\nGiá trị thiếu theo cột:")
print(df.isna().sum())

print("\nSố dòng trùng hoàn toàn:", df.duplicated().sum())
print("Số nội dung text bị trùng:", df.duplicated(subset=["text"]).sum())
```

```python
# Thống kê độ dài văn bản theo số từ
text_length = df["text"].str.split().str.len()

print(text_length.describe().round(2))

df.assign(text_length=text_length) \
  .groupby("label")["text_length"] \
  .describe() \
  .round(2)
```

```python
print(val.isna().sum())
print(val.shape)
label_counts = val["label"].value_counts()
label_ratio = val["label"].value_counts(normalize=True) * 100

print(label_counts)
print()
print(label_ratio.round(2))
```
